```python
from pathlib import Path

DATASET_DIR = Path("../dataset")

print("Dataset path:", DATASET_DIR.resolve())

if DATASET_DIR.exists():
    print("Dataset folder found")
    print("\nContents:")

    for item in DATASET_DIR.iterdir():
        print(" -", item.name)
else:
    print("Dataset folder not found")
```

```python
from pathlib import Path

TRAIN_DIR = DATASET_DIR / "seg_train"
TEST_DIR = DATASET_DIR / "seg_test"

print("TRAIN PATH:", TRAIN_DIR.resolve())
print("TEST PATH :", TEST_DIR.resolve())

print("\nClasses:")
classes = sorted([folder.name for folder in TRAIN_DIR.iterdir() if folder.is_dir()])

for i, class_name in enumerate(classes):
    print(f"{i}: {class_name}")

print(f"\nTotal classes: {len(classes)}")
```
