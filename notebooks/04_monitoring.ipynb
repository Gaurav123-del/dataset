```python
from pathlib import Path
import json

MONITORING_DIR = Path("../monitoring")
MONITORING_DIR.mkdir(parents=True, exist_ok=True)

MONITORING_FILE = MONITORING_DIR / "monitoring_log.json"

if not MONITORING_FILE.exists():
    with open(MONITORING_FILE, "w") as f:
        json.dump([], f, indent=4)

print("Monitoring file:", MONITORING_FILE)
```

```python
from datetime import datetime

def log_prediction(result, response_time):
    with open(MONITORING_FILE, "r") as f:
        logs = json.load(f)

    log_entry = {
        "timestamp": datetime.now().isoformat(),
        "prediction": result["prediction"],
        "confidence": result["confidence"],
        "cached": result["cached"],
        "model": result["model"],
        "response_time_seconds": response_time
    }

    logs.append(log_entry)

    with open(MONITORING_FILE, "w") as f:
        json.dump(logs, f, indent=4)

    print(" Prediction logged")
```

```python
with open(MONITORING_FILE, "r") as f:
    logs = json.load(f)

print("Total Requests:", len(logs))
print(logs)
```

```python
total_requests = len(logs)

cache_hits = sum(
    1 for log in logs
    if log["cached"] is True
)

cache_misses = total_requests - cache_hits

cache_hit_rate = (
    cache_hits / total_requests
    if total_requests > 0
    else 0
)

print("CACHE MONITORING ")
print("Total Requests :", total_requests)
print("Cache Hits     :", cache_hits)
print("Cache Misses   :", cache_misses)
print(f"Cache Hit Rate : {cache_hit_rate:.2%}")
```

```python
response_times = [
    log["response_time_seconds"]
    for log in logs
]

average_response_time = (
    sum(response_times) / len(response_times)
    if response_times
    else 0
)

print("API PERFORMANCE")
print(
    f"Average Response Time: "
    f"{average_response_time:.4f} seconds"
)
```
